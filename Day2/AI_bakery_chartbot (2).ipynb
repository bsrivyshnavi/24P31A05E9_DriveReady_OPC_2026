```python
from langchain_groq import ChatGroq
import os
from google.colab import userdata

print("Libraries imported successfully")
```

```python
os.environ["GROQ_API_KEY"] = userdata.get("Grokapi")

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

print("AI model initialized successfully")
```

```python
response = llm.invoke("How can I order")
print(response.content)
```

```python
SYSTEM = """You are the assistant for Sweet Crumbs Home Bakery.

Facts: we bake to order; orders need 24 hours' notice;
all cakes have eggless versions.

Rules: never promise delivery times;
if you do not know, say so.

Format: reply politely in under 60 words."""
```

```python
def ask(question):
    return llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ]).content
```

```python
print(ask("How can I order?"))
```

```python
print(ask("Do you provide eggless cakes?"))
```

```python
while True:
    question = input("You: ")

    if question.lower() == "exit":
        print("Chatbot: Thank you! Visit again.")
        break

    print("Chatbot:", ask(question))
```
